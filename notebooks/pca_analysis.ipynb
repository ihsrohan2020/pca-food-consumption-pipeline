# PCA Analysis on Food Consumption Data
This notebook loads the cleaned dataset produced by the pipeline, runs PCA, and visualizes consumption patterns across countries.

```python
import pandas as pd
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
from src.clean_transform import run_pipeline

cleaned = run_pipeline("data/food_consumption.csv")
cleaned.head()
```

## Prepare features for PCA

```python
X = cleaned[["consumption", "co2_emission"]]
countries = cleaned["country"]
```

## Run PCA

```python
pca = PCA(n_components=2)
components = pca.fit_transform(X)

pca_df = pd.DataFrame({
    "PC1": components[:, 0],
    "PC2": components[:, 1],
    "country": countries
})

pca_df.head()
```

## Variance explained

```python
explained = pca.explained_variance_ratio_
explained
```

## PCA Scatter Plot

```python
plt.figure(figsize=(8,6))
plt.scatter(pca_df["PC1"], pca_df["PC2"], alpha=0.7)

for i, country in enumerate(pca_df["country"]):
    plt.text(pca_df["PC1"][i] + 0.02, pca_df["PC2"][i] + 0.02, country, fontsize=8)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA of Food Consumption Patterns by Country")
plt.grid(True)
plt.show()
```
